#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
import re

enzR = {
    'HindIII': 'A*AGCTT',
    'BamHI': 'G*GATCC',
    'AluI': 'AG*CT',
    'Sau3AI': '*GATC',
    'EcoRI': 'G*AATTC'
}

def digerir_restriccion(secuencia, enzimas):
    enzimas_con_corte = {}
    enzimas_sin_corte = []
    cortes = []

    for enz, patron in enzimas.items():
        sitio = patron.replace('*', '')
        posicion_rel = patron.find('*')
        posiciones = [m.start() + posicion_rel for m in re.finditer(sitio, secuencia)]
        if posiciones:
            enzimas_con_corte[enz] = posiciones
            cortes.extend(posiciones)
        else:
            enzimas_sin_corte.append(enz)

    cortes_ordenados = sorted(set([0] + cortes + [len(secuencia)]))
    fragmentos = []
    for i in range(len(cortes_ordenados) - 1):
        tamaño = cortes_ordenados[i+1] - cortes_ordenados[i]
        fragmentos.append((i+1, tamaño))

    return {
        'enz_con_corte': enzimas_con_corte,
        'enz_sin_corte': enzimas_sin_corte,
        'fragmentos': fragmentos,
        'num_fragmentos': len(fragmentos)
    }

resultado = digerir_restriccion(secuencia, enzR)

print("1. Enzimas con sitio de corte y posiciones:")
for enz, pos in resultado['enz_con_corte'].items():
    print(f"   {enz}: {pos}")

print("\n2. Enzimas sin sitio de corte:")
for enz in resultado['enz_sin_corte']:
    print(f"   {enz}")

print("\n3. Fragmentos resultantes (número y tamaño):")
for num, tam in resultado['fragmentos']:
    print(f"   Fragmento {num}: {tam} bases")

print(f"\n4. Número total de fragmentos: {resultado['num_fragmentos']}")
```

#### 7.- Realiza una función que mediante la introducción de una secuencia de ADN devuelva lo siguiente: (3pto)
1. Todos los ORFs posibles que puedan surgir de la secuencia.
2. Selecciona los 6 ORFs más largos e indica si corresponden a la strand positiva o negativa.
3. Devuelve como mejor predicción, el más largo de todos.
4. Calcula el porcentaje de aminoácidos aromáticos de esa proteína elegida y su longitud relativa a la secuencia original de ADN.

```python
def reverse_complement(seq):
    return secuencia.translate(str.maketrans('ATCG', 'TAGC'))[::-1]

def encontrar_orfs(seq):
    orfs = []
    stop_codones = ['TAA', 'TAG', 'TGA']
    n = len(secuencia)
    for strand in [1, -1]:
        sec = secuencia if strand == 1 else reverse_complement(secuencia)
        for frame in range(3):
            i = frame
            while i+3 <= n:
                codon = sec[i:i+3]
                if codon == 'ATG':
                    for j in range(i+3, n-2, 3):
                        codon_stop = sec[j:j+3]
                        if codon_stop in stop_codones:
                            orfs.append({
                                'start': i, 'end': j+3, 'length_nt': j+3-i,
                                'strand': '+' if strand==1 else '-',
                                'frame': frame+1,
                                'sec': sec[i:j+3]
                            })
                            break
                i += 3
    return sorted(orfs, key=lambda x: -x['length_nt'])

orfs = encontrar_orfs(secuencia)

# 1. Todos los ORFs posibles (la variable orfs ya los contiene)
print("Cantidad de ORFs encontrados:", len(orfs))

# 2. Los 6 más largos
print("\nLos 6 ORFs más largos:")
for i, orf in enumerate(orfs[:6]):
    print(f"{i+1}. Frame: {orf['frame']}, Strand: {orf['strand']}, Start: {orf['start']}, End: {orf['end']}, Longitud: {orf['length_nt']} nt")

# 3. Mejor predicción y porcentaje relativo
if orfs:
    best_orf = orfs[0]
    print("\nMejor predicción (más largo):")
    print(f"Frame: {best_orf['frame']}, Strand: {best_orf['strand']}, Start: {best_orf['start']}, End: {best_orf['end']}, Longitud: {best_orf['length_nt']} nt")
    porcentaje_longitud = (best_orf['length_nt'] / len(secuencia)) * 100
    print(f"Porcentaje relativo de longitud: {porcentaje_longitud:.2f}%")

    n_codones = best_orf['length_nt'] // 3
    sec_aa = [best_orf['sec'][i:i+3] for i in range(0, len(best_orf['sec']), 3)]
    print(f"\nAminoácidos (simulados): {n_codones}")
    # Porcentaje de cada codón (como "aminoácido simulado")
    from collections import Counter
    conteos = Counter(sec_aa)
    print("\nPorcentaje de cada codón en la proteína más larga:")
    for codon, count in conteos.items():
        print(f"{codon}: {100*count/n_codones:.2f}%")
else:
    print("No se encontraron ORFs.")

```
